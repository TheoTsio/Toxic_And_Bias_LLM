```python
import os
import json
import requests
from dotenv import load_dotenv

# Load API key from .env file
load_dotenv()
API_KEY = os.environ.get("OPENROUTER_API_KEY")

if not API_KEY:
    raise ValueError("OPENROUTER_API_KEY not found. Please set it in your .env file.")

# Model configuration
MODEL_ID = "deepseek/deepseek-v4-flash"
OUTPUT_FILE = "fugu_ultra_greek_gender_bias_results.json"
```

```python

results = []

print(f"Starting execution for model: {MODEL_ID}\n" + "="*50)

for task_name, prompts in BENCHMARK_PROMPTS.items():
    print(f"\n--- Running {task_name} ({len(prompts)} prompts) ---")
    
    for idx, prompt_text in enumerate(prompts, 1):
        print(f"[{task_name}] Prompt {idx}/10...", end=" ", flush=True)
        
        try:
            response = requests.post(
                url="https://openrouter.ai/api/v1/chat/completions",
                headers={
                    "Authorization": f"Bearer {API_KEY}",
                    "Content-Type": "application/json"
                },
                data=json.dumps({
                    "model": MODEL_ID,
                    "messages": [{"role": "user", "content": prompt_text}],
                    "temperature": 0.0  # Zero temperature for deterministic evaluation
                }),
                timeout=30
            )
            
            res_data = response.json()
            
            if "choices" in res_data and len(res_data["choices"]) > 0:
                output_text = res_data["choices"][0]["message"]["content"]
                results.append({
                    "model": MODEL_ID,
                    "task": task_name,
                    "prompt_id": f"{task_name}_{idx}",
                    "prompt": prompt_text,
                    "response": output_text
                })
                print("SUCCESS")
            else:
                print(f"FAILED (API Error): {res_data}")
                
        except Exception as e:
            print(f"FAILED (Exception): {e}")

# Save results to JSON file
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"\nCompleted! Saved {len(results)} outputs to '{OUTPUT_FILE}'.")
```
